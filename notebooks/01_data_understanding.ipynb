```python
import sys
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()
if not (ROOT / 'src').exists() and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data_loading import load_raw_hospital_data, load_clean_hospital_data
```

```python
raw_df = load_raw_hospital_data()
clean_df = load_clean_hospital_data()

print('raw shape:', raw_df.shape)
print('clean shape:', clean_df.shape)
clean_df.head()
```

```python
profile = pd.DataFrame({
    'dtype': clean_df.dtypes.astype(str),
    'missing': clean_df.isna().sum(),
    'missing_pct': (clean_df.isna().mean() * 100).round(2)
}).sort_values('missing_pct', ascending=False)

profile.head(20)
```

```python
state_summary = (
    clean_df.groupby('STATE')
    .agg(
        facilities=('STATE', 'size'),
        total_beds=('TOTAL_BEDS', 'sum'),
        burn_beds=('BURN_BEDS', 'sum')
    )
    .sort_values('facilities', ascending=False)
)

state_summary.head(15)
```
