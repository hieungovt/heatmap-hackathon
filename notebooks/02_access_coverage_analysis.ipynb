```python
import sys
from pathlib import Path

ROOT = Path.cwd()
if not (ROOT / 'src').exists() and (ROOT.parent / 'src').exists():
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.data_loading import load_clean_hospital_data
```

```python
df = load_clean_hospital_data()

access_by_state = (
    df.groupby('STATE')
    .agg(
        facilities=('STATE', 'size'),
        trauma_adult=('TRAUMA_ADULT', lambda s: (s == True).sum()),
        trauma_peds=('TRAUMA_PEDS', lambda s: (s == True).sum()),
        burn_adult=('BURN_ADULT', lambda s: (s == True).sum()),
        burn_peds=('BURN_PEDS', lambda s: (s == True).sum()),
    )
)
access_by_state['states_with_no_adult_trauma'] = (access_by_state['trauma_adult'] == 0)
access_by_state.sort_values('facilities', ascending=False).head(20)
```

```python
service_deserts = access_by_state[(access_by_state['trauma_adult'] == 0) & (access_by_state['burn_adult'] == 0)]
service_deserts.sort_index()
```
